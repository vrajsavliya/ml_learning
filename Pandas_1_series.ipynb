## Series from lists

There are different ways to build the series

```python
# using arrays or can say python list
country =['India','pak','USA','Nepal']
pd.Series(country)
```

```python
#Integer type series
runs = [13,12,3,4,5,62,4,533,5]
pd.Series(runs,dtype=np.int32)
```

```python
# Custom indexing and integer list

marks = [80,99,75,89]
subjects = ['english','math','hindi','SS']

pd.Series(marks,index=subjects) #Custom index bana ne ke liyee best hai yee
```

```python
# setting a name
pd.Series(marks,index=subjects,name='Vraj svliya ke marks')
```

## Series from dict

```python
marks = {
    'maths' : 99,
    'English' : 89,
    'Hindi' : 79,
    'Science' : 86
}

marks_series = pd.Series(marks,name = 'Vraj Savliya ke Marks')
marks_series
```
