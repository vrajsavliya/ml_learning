Numpy array vs Python List

```python
#speed test
#list THIS IS THE LIST IN PYTHON
a = [i for i in range(10000000)]
b = [i for i in range(10000000)]

c = []

import time

start = time.time()
for i in range(len(a)):
  c.append(a[i] + b[i])
print(time.time() - start)
```

```python
# now comparing with the np.array
import numpy as np

a = np.arange(10000000)
b = np.arange(10000000,20000000)
start = time.time()
c = a + b
print(time.time() - start)
```

```python
# now comparing the memory in both list and array.np

a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]
import sys

sys.getsizeof(a)
```

```python
a = np.arange(10000000,dtype = np.int32)
sys.getsizeof(a)
```

Broadcasting

here the rules are for the broadcasting is the dimension should be same as the biggest one

```python
# same shape

a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)

print(a)
print(b)
print(a + b)
```

```python
# diff shape

a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)

print(a)
print(b)
print(a + b)
```

Working with mathematical formulas

```python
import numpy as np

a = np.arange(10)
np.sin(a)
```

```python
# sigmoid
def sigmoid(array):
  return 1/(1 + np.exp(-array))

sigmoid(a)
```

```python
# mean squared error
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
print(actual)
print(predicted)

def mse(actual,predicted):
  return np.mean((actual - predicted)**2)

mse(actual,predicted)
```

```python
# categorical cross entropy

a = np.array([1,1,0,1,0,0])
b = np.array([0.9,0.8,0.3,0.9,0.7,0.6])

def binary(a,b):
  return -np.mean(a*(np.log(b) + (1 - a)*np.log(1 - b)))

binary(a,b)
```
