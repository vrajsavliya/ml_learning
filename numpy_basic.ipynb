ways to create different types of arrays

```python
#np.array ---> use to create a array
import numpy as np

a = np.array([1,2,3]) # This is a 1D array
print(a)
print(type(a))
```

```python
# 2D and 3D array can also be made using the nparray

b = np.array([[1,2,3],[4,5,6]])
print(b)

# tensor is a 3D numpy array

c = np.array([[[1,2],[3,4]],[[2,3],[5,6]]])
print("this is a tensor \n",c)
```

```python
# Lets make a float data type array using np using dtype=float
# can be made the bool and in python all the values are consider true in all the positive numbers
# and complex is the (num + i) yee vala format dekh loo change krr ke dtype me

d = np.array([1,2,3],dtype=bool)
print("float Array",d)
```

```python
# np.arange to use in the loops

np.arange(1,11) # ye deta hai (start, end - 1);
np.arange(1,11,2) # ye 2 hai voo i++ jo haii bus aus ki badle i + 2 karega
```

```python
# with reshape
np.arange(1,11).reshape(5,2)
# here the imp condition for reshape the product sould be equal to the number count then its possible
# ye reshape mujzee 2D me convert krr ke deta hai ye brackets hai(row,colum)
```

```python
# np.ones and np.zeros

# np.ones ko use krr ke hum eek go me sare items array me 1 bana sakte hai
np.ones((3,4))
# same for 0 one use the np.zeros
np.zeros((3,4))

#it will be used in the neurals network to initially give all the values
```

```python
# np.random

np.random.random((3,4))
```

```python
# np.linespace(Start_range,End_range,numbers to generate between this range)
# it generate the points in the equal distance
# this is call linearly seperable
np.linspace(-10,10,10)
```

```python
# np.identity

np.identity(3)
```

Array Attributes

```python
a1 = np.arange(10,dtype=np.int32)
print(a1)
a2 = np.arange(12, dtype=float).reshape(3, 4)
print(a2)
a3 = np.arange(8).reshape(2, 2, 2)
print(a3)
```

```python
# ndim --> tells about the dimensions of the array what it has 1,2,3D etc
a1.ndim
a2.ndim
a3.ndim
```

```python
# shape
# it tell about the numbers of rows col etc the shape of the array

print(a3.shape)
```

```python
# size
# tells about the numbers of items in the array

a1.size
```

```python
# itemsize

a1.itemsize
```

```python
# dtype

a1.dtype
```

Changing Datatype

```python
# astype
a3.astype(np.int32)
```
