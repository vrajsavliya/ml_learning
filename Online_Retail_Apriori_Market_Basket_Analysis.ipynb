```python
print("Cancelled invoices:",
      df["Invoice"].astype(str).str.startswith("C").sum())

print("Negative quantities:",
      (df["Quantity"] < 0).sum())

print("Zero quantities:",
      (df["Quantity"] == 0).sum())

print("Zero/negative prices:",
      (df["Price"] <= 0).sum())

print("Duplicate rows:",
      df.duplicated().sum())
```

```python

df = df[~df["Invoice"].astype(str).str.startswith("C")]

df = df[df["Quantity"] > 0]

df = df[df["Price"] > 0]

df = df.drop_duplicates()

print("Remaining rows:", len(df))
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
print("Unique invoices:", df["Invoice"].nunique())
print("Unique products:", df["Description"].nunique())

print("\nInvoice examples:")
print(df["Invoice"].head(10).to_list())

print("\nProduct examples:")
print(df["Description"].head(10).to_list())
```

```python
df["Description"] = df["Description"].str.strip()
```

```python
print(df["Description"].head(10).to_list())
```

```python
basket = pd.crosstab(
    df["Invoice"],
    df["Description"]
)

basket.head()
```

```python
basket.shape
```

```python
basket.head()
```

```python
basket = (basket > 0).astype(int)
```

```python
basket.head()
```

```python
basket.shape
```

```python
basket["Total_Items"] = basket.sum(axis=1)

basket["Total_Items"].describe()
```

```python
basket = basket.drop(columns=["Total_Items"])
```

```python
product_frequency = basket.sum(axis=0).sort_values(ascending=False)

product_frequency.head(20)
```
